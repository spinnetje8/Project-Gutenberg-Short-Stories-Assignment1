```python
# b) Check that every path in the `file` column exists, and that every file in `data` appears exactly once in the `file` column.
```

```python
files_in_folder = []
for name in os.listdir("data"):
    path = "data/" + name
    if not name.startswith(".") and os.path.isfile(path):
        files_in_folder.append(path)

listed = list(meta["file"])

for path in listed:
    if path not in files_in_folder:
        print("In metadata.csv, but no file:", path)

for path in files_in_folder:
    if listed.count(path) != 1:
        print("Not listed exactly once in metadata.csv:", path)

print("File check finished")
```
