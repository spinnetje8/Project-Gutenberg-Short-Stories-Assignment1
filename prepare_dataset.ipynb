```python
import os

import pandas as pd

meta = pd.read_csv("metadata-draft.csv", encoding = "utf8", sep=";")
meta
```

```python
meta['year_earliest'] = meta['year_earliest'].astype('Int64')

meta['year_latest'] = meta['year_latest'].astype('Int64')

meta
```

```python
ids = []
files = []
sources = []
for nr in meta["ebook_nr"]:
    ids.append("gut_" + str(nr))
    files.append("data/gut_" + str(nr) + ".txt")
    sources.append("https://www.gutenberg.org/ebooks/" + str(nr))

meta["id"] = ids
meta["file"] = files
meta["source"] = sources
meta["licence"] = "Public domain in the USA"    
meta[["id", "file", "source"]]
```

```python
os.makedirs("data", exist_ok=True)

for i in range(len(meta)):
    with open("downloaded/" + meta["original_filename"][i], encoding="utf-8") as f:
        text = f.read()
    with open(meta["file"][i], "w", encoding="utf-8") as f:
        f.write(text)

columns = ["id", "file", "title", "author", "language", "year_earliest", "year_latest", "source", "licence"]
meta[columns].to_csv("metadata.csv", index=False)
meta[columns]
```
